**Mental Health Survey- Machine Learning**

```python
import pandas as pd
import numpy as np
import  seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
import xgboost as xgb
from xgboost import XGBClassifier

from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
df=pd.read_csv("MentalHealth.csv")
```

```python
df.isna().sum()/len(df)
```

```python
df.dropna(inplace=True)
```

```python
df.drop(columns="survey_date",inplace=True)
```

```python
df.info()
```

```python
lst_float=[]

for i in df.select_dtypes("float64"):
  lst_float.append(i)
```

```python
for i in lst_float:
  df[i]=df[i].astype("int64")
```

```python
cat_col=df.select_dtypes(include=object).columns
num_col=df.select_dtypes(exclude=object).columns
```

```python
for i in cat_col:
  df[i]=LabelEncoder().fit_transform(df[i])
```

```python
df["mental_health_risk"].value_counts()
```

```python
x=df.drop(columns="mental_health_risk",axis=1)
y=df["mental_health_risk"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
dtc=DecisionTreeClassifier(max_depth=5,max_features=10)
dtc.fit(x_train,y_train)
```

```python
dtc_pre_tr=dtc.predict(x_train)
dtc_pre_te=dtc.predict(x_test)
```

```python
print("---------Accuracy Score-------------")
dtc_as_tr=dtc.score(x_train,y_train)
print("Training:",dtc_as_tr*100)
dtc_as_te=dtc.score(x_test,y_test)
print("Testing:",dtc_as_te*100,"\n")

print("-----------Confusion Matrix-----------")
dtc_cm_tr=confusion_matrix(y_train,dtc_pre_tr)
print("Training:",dtc_cm_tr)
dtc_cm_te=confusion_matrix(y_test,dtc_pre_te)
print("Testing:",dtc_cm_te,"\n")

print("-----------Classification Report-----------")
dtc_cr_tr=classification_report(y_train,dtc_pre_tr)
print(dtc_cr_tr)
dtc_cr_te=classification_report(y_test,dtc_pre_te)
print(dtc_cr_te)
```

```python
rfc=RandomForestClassifier(max_depth=6)
rfc.fit(x_train,y_train)
```

```python
rfc_pre_tr=rfc.predict(x_train)
rfc_pre_te=rfc.predict(x_test)
```

```python
print("---------Accuracy Score-------------")
rfc_as_tr=rfc.score(x_train,y_train)
print("Training:",rfc_as_tr*100)
rfc_as_te=rfc.score(x_test,y_test)
print("Testing:",rfc_as_te*100,"\n")

print("-----------Confusion Matrix-----------")
rfc_cm_tr=confusion_matrix(y_train,rfc_pre_tr)
print("Training:",rfc_cm_tr)
rfc_cm_te=confusion_matrix(y_test,rfc_pre_te)
print("Testing:",rfc_cm_te,"\n")

print("-----------Classification Report-----------")
rfc_cr_tr=classification_report(y_train,rfc_pre_tr)
print(rfc_cr_tr)
rfc_cr_te=classification_report(y_test,rfc_pre_te)
print(rfc_cr_te)
```

```python
lr = LogisticRegression(max_iter=100)
lr.fit(x_train, y_train)
```

```python
lr_pre_tr = lr.predict(x_train)
lr_pre_te = lr.predict(x_test)
```

```python
print("---------Accuracy Score-------------")
lr_as_tr = lr.score(x_train, y_train)
print("Training:", lr_as_tr * 100)
lr_as_te = lr.score(x_test, y_test)
print("Testing:", lr_as_te * 100, "\n")

print("-----------Confusion Matrix-----------")
lr_cm_tr = confusion_matrix(y_train, lr_pre_tr)
print("Training:", lr_cm_tr)
lr_cm_te = confusion_matrix(y_test, lr_pre_te)
print("Testing:", lr_cm_te, "\n")

print("-----------Classification Report-----------")
lr_cr_tr = classification_report(y_train, lr_pre_tr)
print(lr_cr_tr)
lr_cr_te = classification_report(y_test, lr_pre_te)
print(lr_cr_te)
```

```python
xgb = XGBClassifier(max_depth=6, eval_metric='logloss')
xgb.fit(x_train, y_train)
```

```python
xgb_pre_tr = xgb.predict(x_train)
xgb_pre_te = xgb.predict(x_test)
```

```python
print("---------Accuracy Score-------------")
xgb_as_tr = xgb.score(x_train, y_train)
print("Training:", xgb_as_tr * 100)
xgb_as_te = xgb.score(x_test, y_test)
print("Testing:", xgb_as_te * 100, "\n")

print("-----------Confusion Matrix-----------")
xgb_cm_tr = confusion_matrix(y_train, xgb_pre_tr)
print("Training:", xgb_cm_tr)
xgb_cm_te = confusion_matrix(y_test, xgb_pre_te)
print("Testing:", xgb_cm_te, "\n")

print("-----------Classification Report-----------")
xgb_cr_tr = classification_report(y_train, xgb_pre_tr)
print(xgb_cr_tr)
xgb_cr_te = classification_report(y_test, xgb_pre_te)
print(xgb_cr_te)
```
